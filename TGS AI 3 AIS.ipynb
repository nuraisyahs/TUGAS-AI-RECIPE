## 1. Recipe Audit Station CSV

### Problem

Baca satu file CSV stasiun dan identifikasi masalah pada skema kolom,
rentang waktu, nilai hilang, serta timestamp duplikat. Hasilnya berupa
ringkasan QA/QC, bukan analisis iklim atau prediksi hujan. File sumber
tidak boleh diubah.

### Minimal Theory

Audit data dilakukan sebelum agregasi atau analisis agar kesalahan
struktur, waktu, dan nilai dapat diketahui lebih awal. Timestamp ganda
dapat memengaruhi agregasi, sedangkan nilai hilang atau tipe kolom yang
tidak sesuai dapat mengubah hasil analisis.

### Input dan Constraints

-   `CSV_PATH`: path file CSV.
-   `TIME_COLUMN`: nama kolom waktu.
-   `UNITS`: kontrak satuan untuk kolom yang relevan.
-   Pertahankan file sumber; gunakan pembacaan *read-only*.
-   Jangan menghapus atau memperbaiki baris sumber secara otomatis.
-   Jangan menyimpulkan hasil ilmiah yang tidak diuji oleh audit.

### Prompt dan Procedure

> Buat draft recipe untuk audit QA/QC station CSV. Periksa skema kolom,
> rentang waktu, nilai hilang, timestamp duplikat, dan tipe data tanpa
> mengubah input mentah. Nyatakan input, teori minimum, asumsi,
> prosedur, output yang diharapkan, rencana implementasi Python,
> pemeriksaan kegagalan, serta status eksekusi. Pisahkan spesifikasi
> recipe dari klaim bahwa program telah dijalankan.

Langkah kerja: 1. Baca CSV secara *read-only*. 2. Periksa apakah kolom
wajib tersedia. 3. Validasi tipe data dan rentang timestamp. 4. Hitung
nilai hilang dan timestamp duplikat. 5. Buat laporan QA/QC dan catatan
pemeriksaan. 6. Catat error atau kolom yang tidak sesuai tanpa mengubah
sumber.

### Code / Implementation Plan

Implementasi dapat menggunakan Python dengan `pandas` untuk membaca CSV
dan menghitung ringkasan. Kode disimpan pada path relatif, misalnya
`code/recipes/io/r-io-001.py`. Path ini merupakan rancangan, bukan bukti
bahwa file sudah dibuat.

### Output

-   `results/io/station-summary.md`: laporan Markdown.
-   `results/run-records/r-io-001.json`: run record jika program
    dijalankan.

Laporan mencakup jumlah baris dan kolom, kolom yang ditemukan, rentang
waktu, jumlah nilai hilang, jumlah timestamp duplikat, dan status setiap
pemeriksaan.

### Optional Check / Evidence

Periksa bahwa file input tidak berubah, laporan tidak kosong, jumlah
baris sesuai dengan sumber, dan status eksekusi dicatat. Gunakan
`Not yet executed` jika program belum dijalankan. Label `Executed` hanya
digunakan jika bukti eksekusi tersedia.

## 2. Recipe Normalisasi GeoJSON

### Problem

Susun recipe untuk mengubah file GeoJSON menjadi GeoJSON yang
dinormalisasi, inventaris CSV, dan ringkasan JSON. Recipe harus
memeriksa geometri, properti, geometri kosong, serta urutan koordinat
tanpa menebak informasi yang tidak tersedia.

### Minimal Theory

GeoJSON menggunakan struktur JSON untuk menyimpan fitur geografis.
Pemeriksaan perlu mencakup struktur `FeatureCollection`, tipe geometri,
koordinat, properti, dan sistem referensi koordinat. Untuk koordinat
geografis GeoJSON, urutan koordinat umumnya berupa bujur lalu lintang.

### Input dan Constraints

-   `GEOJSON_PATH`: path GeoJSON sumber.
-   Path output untuk GeoJSON, CSV inventaris, dan JSON ringkasan.
-   Jangan menghilangkan fitur atau properti tanpa aturan yang jelas.
-   Jangan menebak CRS atau mengubah urutan koordinat tanpa dasar.
-   Keberhasilan membaca file tidak otomatis membuktikan topologi
    geometri valid.

### Prompt dan Procedure

> Susun recipe untuk menormalisasi `[GEOJSON_PATH]` menjadi GeoJSON,
> inventaris CSV, dan ringkasan JSON. Periksa struktur, tipe geometri,
> properti, geometri kosong atau tidak valid, urutan koordinat, dan
> informasi CRS. Nyatakan ketidakpastian secara terbuka dan jangan
> mengklaim output telah dihasilkan sebelum eksekusi dibuktikan.

Langkah kerja: 1. Baca dan validasi sintaks JSON. 2. Periksa struktur
objek dan daftar fitur. 3. Periksa geometri, koordinat, serta properti.
4. Identifikasi CRS atau urutan koordinat yang belum jelas. 5.
Normalisasi hanya berdasarkan aturan yang terdefinisi. 6. Ekspor
inventaris fitur dan ringkasan hasil pemeriksaan. 7. Catat fitur yang
gagal diproses.

### Code / Implementation Plan

Implementasi dapat menggunakan Python dengan pustaka yang mendukung
GeoJSON, misalnya `json` untuk pemeriksaan struktur dan pustaka
geospasial yang tersedia di lingkungan kerja. Dependency harus dicatat
dan tidak boleh diasumsikan sudah terpasang.

### Output

-   GeoJSON hasil normalisasi.
-   CSV inventaris fitur.
-   JSON ringkasan QA/QC.
-   Run record serta peringatan jika tersedia.

### Optional Check / Evidence

Bandingkan jumlah fitur sumber dan hasil, periksa properti penting,
validasi struktur output, dan catat fitur yang tidak berhasil diproses.
Jika program belum dijalankan, sebutkan output sebagai hasil yang
direncanakan.

## 3. Recipe Model Difusi Panas Satu Dimensi

### Problem

Buat recipe untuk mensimulasikan perubahan suhu terhadap posisi dan
waktu berdasarkan profil awal, difusivitas termal, kondisi batas, jarak
grid, dan langkah waktu.

### Minimal Theory

Persamaan difusi panas satu dimensi adalah:

\[
`\frac{\partial T}{\partial t}`{=tex}=`\alpha`{=tex}`\frac{\partial^2T}{\partial x^2}`{=tex}
\]

dengan (T) adalah suhu, (t) waktu, (x) posisi, dan (`\alpha`{=tex})
difusivitas termal.

Untuk skema beda hingga eksplisit standar, parameter kestabilan dapat
ditulis:

\[ r=`\frac{\alpha\Delta t}{(\Delta x)^2}`{=tex} \]

Pada kasus satu dimensi dengan skema eksplisit standar, syarat
kestabilan yang umum digunakan adalah (r`\leq 1`{=tex}/2). Syarat ini
perlu disesuaikan dengan skema numerik yang benar-benar digunakan.

### Input dan Constraints

-   `INITIAL_PROFILE`: profil suhu awal.
-   `ALPHA`: difusivitas termal dan satuannya.
-   `BOUNDARIES`: kondisi batas.
-   `DZ`: jarak grid.
-   `DT`: langkah waktu.
-   Pastikan konsistensi satuan.
-   Jangan mengklaim kalibrasi atau konvergensi tanpa pengujian.

### Prompt dan Procedure

> Buat draft recipe model difusi panas satu dimensi menggunakan profil
> awal `[INITIAL_PROFILE]`, difusivitas `[ALPHA]`, kondisi batas
> `[BOUNDARIES]`, jarak grid `[DZ]`, dan langkah waktu `[DT]`. Sertakan
> persamaan pengatur, konvensi satuan dan tanda, pemeriksaan kestabilan,
> pembanding analitik atau kasus dasar, profil output, grafik, run
> record, serta peringatan jika parameter atau dependency belum
> tersedia. Jangan menyatakan eksekusi tanpa bukti.

Langkah kerja: 1. Validasi parameter dan satuan. 2. Bentuk grid ruang
dan waktu. 3. Terapkan kondisi awal serta kondisi batas. 4. Periksa
kestabilan skema. 5. Jalankan iterasi jika implementasi tersedia. 6.
Bandingkan hasil dengan kasus dasar atau solusi analitik jika
memungkinkan. 7. Simpan profil suhu, grafik, dan catatan eksekusi.

### Code / Implementation Plan

Implementasi dapat menggunakan Python, `numpy` untuk komputasi numerik,
dan `matplotlib` untuk visualisasi. Pilihan metode numerik, parameter,
serta dependency harus dicatat.

### Output

-   Profil suhu terhadap posisi dan waktu.
-   Grafik hasil simulasi.
-   Ringkasan parameter.
-   Run record berisi status dan peringatan.

### Optional Check / Evidence

Periksa kestabilan numerik, kondisi batas, konsistensi satuan, dan
kesesuaian dengan kasus dasar. Jangan mengklaim konvergensi, kalibrasi,
atau keberhasilan eksekusi tanpa bukti yang sesuai.

# BAGIAN II --- RECIPE BARU BUATAN SAYA

## 4. Recipe QC dan Analisis Waktu Tempuh Seismik Refraksi

### Problem

Membuat workflow yang dapat diulang untuk memeriksa kualitas data waktu
tiba gelombang P pada survei seismik refraksi, memvisualisasikan
hubungan waktu tempuh dengan jarak sumber--geofon, dan memperkirakan
kecepatan semu jika data memenuhi syarat.

Masalah yang hendak diperiksa meliputi nilai hilang, satuan tidak
konsisten, jarak yang tidak valid, duplikasi pasangan sumber--geofon,
serta waktu tiba yang tidak masuk akal. Tujuannya adalah menghasilkan
analisis awal yang dapat diperiksa sebelum interpretasi struktur bawah
permukaan dilakukan.

### Minimal Theory

Dalam pendekatan sederhana, hubungan waktu tempuh terhadap jarak dapat
dituliskan:

\[ t(x)=t_0+`\frac{x}{v}`{=tex} \]

dengan (t(x)) waktu tempuh, (t_0) intersep waktu, (x) jarak
sumber--geofon, dan (v) kecepatan.

Kemiringan grafik waktu tempuh terhadap jarak adalah:

\[ m=`\frac{\Delta t}{\Delta x}`{=tex} \]

Jika hubungan linear layak digunakan, kecepatan semu dapat diperkirakan
dari:

\[ v\_{`\mathrm{app}`{=tex}}=`\frac{1}{m}`{=tex} \]

Untuk waktu dalam sekon dan jarak dalam meter, satuan kecepatan adalah
m/s. Jika waktu tercatat dalam milisekon, lakukan konversi terlebih
dahulu.

Hubungan linear merupakan pendekatan awal. Pada data nyata, beberapa
segmen dapat menunjukkan gelombang yang merambat melalui lapisan
berbeda. Kecepatan semu tidak otomatis sama dengan kecepatan intrinsik
batuan dan tidak cukup untuk memastikan jenis batuan.

### Input dan Constraints

Input utama: - `DATA_PATH`: lokasi CSV data seismik. - `SHOT_ID`:
identitas sumber/tembakan. - `GEOPHONE_ID`: identitas geofon. -
`DISTANCE`: jarak sumber--geofon. - `P_ARRIVAL_TIME`: waktu tiba
gelombang P. - `DISTANCE_UNIT`: satuan jarak. - `TIME_UNIT`: satuan
waktu.

Input tambahan jika tersedia: - `LINE_ID`: identitas lintasan. -
`PICK_QUALITY`: indikator kualitas picking. - `ELEVATION`: elevasi
sumber atau geofon. - `NOTES`: catatan lapangan.

Batasan: 1. File mentah harus dipertahankan dan dibaca secara
*read-only*. 2. Nilai hilang tidak boleh diisi dengan perkiraan tanpa
penandaan. 3. Baris bermasalah harus dilaporkan, bukan dihapus
diam-diam. 4. Pasangan sumber--geofon yang terduplikasi harus diperiksa.
5. Jarak dan waktu harus dikonversi ke satuan yang konsisten. 6. Fitting
hanya dilakukan jika jumlah dan sebaran data cukup. 7. Jangan
menyimpulkan litologi hanya berdasarkan nilai kecepatan.

### Prompt dan Procedure

**Prompt yang dapat digunakan kembali:**

> Anda berperan sebagai asisten pengolahan data geofisika untuk analisis
> awal seismik refraksi. Baca `[DATA_PATH]` secara read-only. Periksa
> kolom wajib, tipe data, satuan, nilai hilang, nilai nonnumerik, jarak
> negatif, duplikasi pasangan sumber--geofon, dan konsistensi waktu tiba
> gelombang P.
>
> Buat laporan QA/QC sebelum melakukan perhitungan. Jika data memadai,
> buat grafik waktu tiba P terhadap jarak sumber--geofon. Lakukan
> fitting linear atau analisis segmen hanya jika didukung sebaran data.
> Laporkan metode, kemiringan, intersep, jumlah titik yang digunakan,
> satuan, dan residual atau ukuran kecocokan yang sesuai. Hitung
> kecepatan semu hanya jika kemiringan dapat ditafsirkan dengan benar
> dan satuannya konsisten.
>
> Pisahkan fakta dari data, hasil perhitungan, asumsi, dan interpretasi.
> Jangan menetapkan jenis batuan atau struktur bawah permukaan secara
> pasti dari fitting sederhana. Tentukan output yang diharapkan, simpan
> laporan dan grafik pada path yang dinyatakan, serta buat run record
> hanya jika program benar-benar dijalankan. Jika belum dieksekusi,
> tulis `Not yet executed`.

**Procedure:** 1. Baca data dan catat identitas input. 2. Validasi
kolom, tipe data, satuan, dan rentang nilai. 3. Hitung baris valid,
nilai hilang, duplikasi, dan baris bermasalah. 4. Konversi jarak dan
waktu ke satuan yang konsisten. 5. Buat grafik waktu tiba P terhadap
jarak. 6. Tinjau apakah pola data mendukung fitting linear atau beberapa
segmen. 7. Hitung parameter fitting dan kecepatan semu jika layak. 8.
Evaluasi residual atau ukuran kecocokan model. 9. Buat laporan QA/QC,
tabel hasil, dan grafik. 10. Catat status eksekusi serta keterbatasan
interpretasi.

### Code / Implementation Plan

Rencana implementasi menggunakan Python: - `pandas`: membaca dan
memeriksa data tabular. - `numpy`: perhitungan numerik dan fitting yang
sesuai. - `matplotlib`: grafik waktu tempuh terhadap jarak. - `json` dan
`pathlib`: metadata, path output, dan run record.

Rancangan direktori:

``` text
seismic-refraction-qc/
├── data/
│   └── refraction-example.csv
├── code/
│   └── refraction_qc.py
└── results/
    ├── refraction-qc-report.md
    ├── travel-time-plot.png
    ├── travel-time-results.csv
    └── run-record.json
```

Struktur tersebut merupakan rancangan. Keberadaan file dan keberhasilan
program harus diverifikasi setelah implementasi.

### Output

1.  `refraction-qc-report.md`: ringkasan input, jumlah data, kolom,
    nilai hilang, duplikasi, satuan, dan peringatan.
2.  `travel-time-plot.png`: grafik waktu tiba P terhadap jarak dengan
    label sumbu dan satuan.
3.  `travel-time-results.csv`: jarak, waktu observasi, waktu prediksi
    jika ada, residual, dan status QC.
4.  `run-record.json`: perintah, status keluar, path input/output,
    peringatan, dan informasi lingkungan jika tersedia.

Output di atas merupakan target yang diharapkan, bukan klaim bahwa file
tersebut telah dibuat.

### Optional Check / Evidence

Checklist pemeriksaan: - \[ \] Input berhasil dibaca dan tidak diubah. -
\[ \] Kolom wajib dan satuan telah diperiksa. - \[ \] Nilai hilang,
duplikasi, dan nilai tidak valid dilaporkan. - \[ \] Grafik memiliki
label sumbu dan satuan. - \[ \] Metode fitting dan jumlah titik
dijelaskan. - \[ \] Kecepatan semu dihitung dengan satuan konsisten. -
\[ \] Residual atau ukuran kecocokan ditinjau jika fitting dilakukan. -
\[ \] Interpretasi dibedakan dari hasil perhitungan. - \[ \] Run record
dan status eksekusi didukung bukti.

Status yang boleh digunakan: - `Draft`: spesifikasi recipe sedang
disusun. - `Not yet executed`: program belum dijalankan. - `Partial`:
sebagian langkah selesai, tetapi sebagian lain belum selesai atau belum
diperiksa. - `Executed`: program dijalankan dan bukti eksekusi tersedia.

### Keterbatasan dan Risiko

Hasil bergantung pada kualitas picking, sebaran jarak, geometri survei,
topografi, dan kondisi bawah permukaan. Fitting linear sederhana dapat
tidak mewakili semua gelombang yang tercatat. Kecepatan semu harus
diperlakukan sebagai hasil analisis awal dan bukan bukti tunggal untuk
identifikasi batuan atau penentuan model bawah permukaan.

### Ringkasan Recipe Baru

Recipe ini membuat proses analisis awal seismik refraksi lebih
terstruktur dan dapat diulang. Alurnya dimulai dari pemeriksaan data,
dilanjutkan visualisasi dan estimasi parameter, lalu diakhiri dengan
pemeriksaan hasil serta dokumentasi bukti. Fokus utamanya bukan hanya
memperoleh angka kecepatan, melainkan memastikan angka tersebut dapat
ditelusuri ke data, satuan, metode, dan asumsi yang jelas.
